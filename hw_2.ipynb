```python
import matplotlib.pyplot as plt
import numpy as np
```

```python
ages = np.random.randint(18, 66, 100)
mean_age = np.mean(ages)
```

```python
plt.figure(figsize=(8, 5))
plt.hist(ages, bins=10, color='skyblue', edgecolor='black', alpha=0.7)

# Додавання лінії середнього значення
plt.axvline(mean_age, color='red', linestyle='dashed', linewidth=2, label=f'Середнє: {mean_age:.1f}')

plt.title('Розподіл вікових груп')
plt.xlabel('Вік')
plt.ylabel('Кількість осіб')
plt.legend()
plt.show()
```
