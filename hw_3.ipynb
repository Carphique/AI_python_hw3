```python
import matplotlib.pyplot as plt
import numpy as np
```

```python
group1 = np.random.normal(70, 10, 30)
group2 = np.random.normal(75, 8, 30)
group3 = np.random.normal(65, 12, 30)
```

```python
plt.figure(figsize=(8, 5))
plt.boxplot([group1, group2, group3], tick_labels=['Група 1', 'Група 2', 'Група 3'], patch_artist=True)

plt.title('Розкид результатів іспитів між групами')
plt.xlabel('Групи студентів')
plt.ylabel('Оцінки')
plt.grid(axis='y', linestyle='--', alpha=0.7)
plt.show()
```
