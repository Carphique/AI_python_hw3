```python
import matplotlib.pyplot as plt
```

```python
hours = list(range(24))
load = [20, 18, 15, 15, 17, 30, 50, 75, 85, 90, 88, 85, 80, 82, 85, 90, 95, 90, 80, 60, 45, 35, 25, 22]
```

```python
plt.figure(figsize=(10, 5))
plt.plot(hours, load, color='purple', label='Навантаження')

# Зафарбовування області
plt.fill_between(hours, load, color='purple', alpha=0.3)

plt.title('Погодинне навантаження сервера за добу')
plt.xlabel('Година доби')
plt.ylabel('Навантаження (%)')
plt.xticks(hours) # Показуємо кожну годину на осі X
plt.grid(True, linestyle='--', alpha=0.6)
plt.legend()
plt.show()
```
