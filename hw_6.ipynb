```python
import matplotlib.pyplot as plt
```

```python
months = ['Січ', 'Лют', 'Бер', 'Кві', 'Тра', 'Чер']
conversion = [2.1, 2.3, 2.2, 2.5, 2.8, 3.0]
retention = [40, 42, 45, 44, 48, 50]
avg_check = [120, 125, 130, 128, 135, 140]
orders = [100, 110, 105, 120, 130, 145]
```

```python
fig, axs = plt.subplots(2, 2, figsize=(12, 8))
fig.suptitle('Метрики продукту за перше півріччя', fontsize=16)

# Графік 1: Конверсія
axs[0, 0].plot(months, conversion, color='blue', marker='o')
axs[0, 0].set_title('Конверсія (%)')

# Графік 2: Утримання
axs[0, 1].plot(months, retention, color='green', marker='s')
axs[0, 1].set_title('Утримання (%)')

# Графік 3: Середній чек
axs[1, 0].plot(months, avg_check, color='orange', marker='^')
axs[1, 0].set_title('Середній чек ($)')

# Графік 4: Кількість замовлень
axs[1, 1].plot(months, orders, color='red', marker='d')
axs[1, 1].set_title('Кількість замовлень (тис.)')

# Загальні налаштування для всіх 4 графіків
for ax in axs.flat:
    ax.grid(True, linestyle='--', alpha=0.5)
    ax.set_xlabel('Місяці')

plt.tight_layout() # Автоматично вирівнює відступи між графіками
plt.show()
```
