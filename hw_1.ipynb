```python
import matplotlib.pyplot as plt
```

```python
months = ['Січ', 'Лют', 'Бер', 'Кві', 'Тра', 'Чер', 'Лип', 'Сер', 'Вер', 'Жов', 'Лис', 'Гру']
plan = [100, 120, 130, 110, 150, 160, 170, 180, 200, 210, 230, 250]
fact = [90, 110, 140, 105, 145, 165, 160, 185, 210, 205, 240, 260]
```

```python
plt.figure(figsize=(10, 5))
plt.plot(months, plan, label='План', marker='o', linestyle='--')
plt.plot(months, fact, label='Факт', marker='s')

plt.title('Планові та фактичні продажі за місяцями')
plt.xlabel('Місяці')
plt.ylabel('Продажі (шт.)')
plt.legend()
plt.grid(True)
plt.show()
```
