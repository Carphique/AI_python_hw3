```python
import matplotlib.pyplot as plt
```

```python
days = ['2023-10-01', '2023-10-02', '2023-10-03', '2023-10-04', '2023-10-05', '2023-10-06', '2023-10-07']
temp = [20, 22, 21, 24, 25, 23, 20]
humidity = [60, 55, 58, 50, 48, 52, 65]
```

```python
plt.figure(figsize=(9, 5))
plt.plot(days, temp, label='Температура (°C)', color='red', marker='o')
plt.plot(days, humidity, label='Вологість (%)', color='blue', marker='s')

plt.title('Динаміка температури та вологості за тиждень')
plt.xlabel('Дні')
plt.ylabel('Значення')
plt.xticks(rotation=45) # Поворот підписів осі X
plt.legend()
plt.grid(True)
plt.tight_layout() # Щоб підписи не обрізалися
plt.show()
```
